# DATA TRANSFORM

````python
data = data.sort_values(by=['Date', 'Name_Company', 'Envent', 'Total Revenue'], ascending=False).drop_duplicates(subset=['Date', 'Name_Company', 'Envent'], keep='first')
````

````python
data = data.sort_values(by=['Date', 'Name_Company', 'Envent'], ascending=False).drop_duplicates(subset=['Date', 'Name_Company'], keep='first')
````

````python
data.columns
````

````python
# Các cột numeric (loại bỏ string, label, etc.)
exclude_cols = ["Name_Company", "Date", "Etablishment year", "Event time", "Time", "Envent"]
num_cols = [c for c in data.columns if c not in exclude_cols]
````

````python
data.shape
````

````python
# 1. Sinh lag features
for col in num_cols:
    for lag in [1, 2, 3]:
        data[f"{col}_lag{lag}"] = data.groupby("Name_Company")[col].shift(lag)
````

````python
# Lấy mean, std
for col in num_cols:
    for window in [3, 6]:
        data[f"{col}_rollmean{window}"] = (
            data.groupby("Name_Company")[col].transform(lambda x: x.rolling(window).mean())
        )
        data[f"{col}_rollstd{window}"] = (
            data.groupby("Name_Company")[col].transform(lambda x: x.rolling(window).std())
        )
````

````python
print(data.columns)
````

````python
# 3. Time-based features
data["year"] = data["Date"].dt.year
data["month"] = data["Date"].dt.month
data["quarter"] = data["Date"].dt.quarter

data_for_ts = data.drop(columns=["Name_Company", "Date", "Etablishment year", "Event time", "Time"])
# 4. Xử lý missing do lag/rolling
data_for_ts = data_for_ts.dropna().reset_index(drop=True)
````

````python
# 5. Chuẩn bị X, y
X = data_for_ts.drop(columns="Envent")   # features
y = data_for_ts["Envent"]                  # label
````

````python
X = X.reset_index(drop=True)
y = y.reset_index(drop=True)
````

````python
y.value_counts()/len(y)
````

# MODELING

````python
tscv = TimeSeriesSplit(n_splits=5)
````

````python
from IPython.display import display, Markdown
display(Markdown("![Logo](https://towardsdatascience.com/wp-content/uploads/2024/01/1vSN6l1gSt1UG-MrRFmLx2Q.png)"))
````

````python
acc_scores, f1_scores, auc_scores = [], [], []
````

````python
for fold, (train_idx, test_idx) in enumerate(tscv.split(X)):
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    # Logistic Regression (có penalty L2) L2 → co tất cả hệ số lại gần 0 nhưng vẫn giữ lại tất cả.
    model = LogisticRegression(max_iter=1000, penalty='l2')
    model.fit(X_train, y_train)

    # Dự đoán
    y_pred = model.predict(X_test)
    y_proba = model.predict_proba(X_test)[:, 1]

    # Đánh giá
    acc = accuracy_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    try:
        auc = roc_auc_score(y_test, y_proba)
    except ValueError:
        auc = 0.5 # random guess
    acc_scores.append(acc)
    f1_scores.append(f1)
    auc_scores.append(auc)

    print(f"Fold {fold+1}: Accuracy={acc:.3f}, F1={f1:.3f}, AUC={auc:.3f}")
    print(classification_report(y_test, y_pred))
    cm = confusion_matrix(y_test, y_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=[0, 1])

    disp.plot(cmap=plt.cm.Blues)
    plt.title("Confusion Matrix")
    plt.show()

print("Trung bình:")
print("Accuracy:", np.mean(acc_scores))
print("F1:", np.mean(f1_scores))
print("AUC:", np.mean(auc_scores))
````

```
F1: trung bình điều hòa giữa Precision và Recall
Nếu Precision cao nhưng Recall thấp → F1 thấp.
Nếu Recall cao nhưng Precision thấp → F1 thấp.
Chỉ khi cả hai đều cân bằng, F1 mới cao.
---> Ở đây F1 đều nhỏ, có thể là do dữ liệu bị imbalance
```

````python
cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=[0, 1])

disp.plot(cmap=plt.cm.Blues)
plt.title("Confusion Matrix")
plt.show()
````

**Logistic Regression cho TimeSeries**

```
- Về giả định:
+ Dữ liệu time series không độc lập (giá trị tại thời điểm t có thể phụ thuộc vào t-1, t-2,...)
+ Có thể có tự tương quan (autocorrelation) hoặc chu kỳ (seasonality).
-> cần phải biến đổi dữ liệu để logistic regression dùng được bằng cách tạo lag features. Ngoài ra có thể thêm rolling statistics: mean, std, volatility trong cửa sổ thời gian.
- Về cách chia dữ liệu
+ Không thể shuffle ngẫu nhiên (sẽ phá cấu trúc thời gian).

Xuất phát từ mô hình GLM
